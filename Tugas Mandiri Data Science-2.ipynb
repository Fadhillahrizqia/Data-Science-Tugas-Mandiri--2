## 0. Setup & Load Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('/kaggle/input/datasets/toramky/automobile-dataset/Automobile_data.csv')
df.head()
```

## 1. Data Preparation – Data Cleansing

Data Cleansing terdiri dari:
1. Menentukan tipe data yang benar
2. Menangani missing value (drop atau replace)
3. Menangani outlier

### 1.1 Mengganti '?' menjadi NaN, lalu cek tipe data

Missing value di dataset ini ditulis sebagai `'?'`, bukan `NaN`, sehingga perlu dikonversi dulu sebelum ditangani.

```python
# Ganti semua '?' dengan NaN
df = df.replace('?', np.nan)

# Cek missing value per kolom
print("Missing value per kolom:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df.dtypes
```

**Interpretasi:** Beberapa kolom yang seharusnya numerik (`normalized-losses`, `bore`, `stroke`, `horsepower`, `peak-rpm`, `price`) masih bertipe `object` karena sebelumnya tercampur string `'?'`. Kolom-kolom ini dikonversi ke tipe numerik **setelah** missing value ditangani.

### 1.2 Konversi tipe data numerik yang salah

```python
num_cols_to_fix = ['normalized-losses', 'bore', 'stroke', 'horsepower', 'peak-rpm', 'price']
for col in num_cols_to_fix:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[num_cols_to_fix].dtypes
```

### 1.3 Summary Data (tipe, nilai unik, missing, contoh nilai)

```python
unique_vals = df.nunique()
missing_vals = df.isnull().sum()
data_types = df.dtypes
sample_vals = df.apply(lambda x: x.dropna().unique()[:5])

summary = pd.DataFrame({
    "Tipe Data": data_types,
    "Nilai Unik": unique_vals,
    "Missing": missing_vals,
    "Contoh Nilai": sample_vals
})
summary
```

## 2. Mini Task – Data Cleansing

Instruksi mini task:
1. Perbaiki atribut **normalized-losses** dengan modus.
2. Perbaiki atribut **bore** dengan mean.
3. Perbaiki atribut **stroke** dengan mean.
4. Perbaiki **horsepower** dengan median.
5. Perbaiki **peak-rpm** dengan modus.
6. Perbaiki **price** dengan menghapus baris yang NaN.
7. Perbaiki **num-of-doors** dengan menghapus baris yang NaN.

### 2.1 normalized-losses → diisi modus

```python
mode_normalized_losses = df['normalized-losses'].mode()[0]
df['normalized-losses'] = df['normalized-losses'].fillna(mode_normalized_losses)
print("Modus normalized-losses:", mode_normalized_losses)
print("Sisa missing value:", df['normalized-losses'].isnull().sum())
```

### 2.2 bore → diisi mean

```python
mean_bore = df['bore'].mean()
df['bore'] = df['bore'].fillna(mean_bore)
print("Mean bore:", mean_bore)
print("Sisa missing value:", df['bore'].isnull().sum())
```

### 2.3 stroke → diisi mean

```python
mean_stroke = df['stroke'].mean()
df['stroke'] = df['stroke'].fillna(mean_stroke)
print("Mean stroke:", mean_stroke)
print("Sisa missing value:", df['stroke'].isnull().sum())
```

### 2.4 horsepower → diisi median

```python
median_horsepower = df['horsepower'].median()
df['horsepower'] = df['horsepower'].fillna(median_horsepower)
print("Median horsepower:", median_horsepower)
print("Sisa missing value:", df['horsepower'].isnull().sum())
```

### 2.5 peak-rpm → diisi modus

```python
mode_peak_rpm = df['peak-rpm'].mode()[0]
df['peak-rpm'] = df['peak-rpm'].fillna(mode_peak_rpm)
print("Modus peak-rpm:", mode_peak_rpm)
print("Sisa missing value:", df['peak-rpm'].isnull().sum())
```

### 2.6 price → baris dengan NaN dihapus

```python
before = df.shape[0]
df = df.dropna(subset=['price'])
after = df.shape[0]
print(f"Jumlah baris sebelum: {before}, sesudah: {after}, dihapus: {before - after}")
```

### 2.7 num-of-doors → baris dengan NaN dihapus

```python
before = df.shape[0]
df = df.dropna(subset=['num-of-doors'])
after = df.shape[0]
print(f"Jumlah baris sebelum: {before}, sesudah: {after}, dihapus: {before - after}")
```

```python
# Cek ulang: pastikan semua missing value sudah tertangani
df.isnull().sum()
```

**Interpretasi Mini Task:** Untuk kolom numerik dengan distribusi relatif normal (`bore`, `stroke`), **mean** dipakai. Untuk kolom yang lebih diskrit/berpotensi skewed (`normalized-losses`, `peak-rpm`), **modus** dipakai. `horsepower` diisi **median** karena lebih tahan terhadap outlier. Sementara `price` dan `num-of-doors` yang missing-nya sedikit dan merupakan variabel penting, baris tersebut langsung **dihapus** agar tidak memasukkan bias dari nilai buatan.

## 3. Menangani Outlier

### 3.1 Identifikasi outlier tiap kolom numerik (IQR, threshold 1.5)

```python
exclude_col = ['symboling', 'price']

for col in df.columns:
    if df[col].dtypes in [np.int64, np.float64] and col not in exclude_col:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        col_min = df[col].min()
        col_max = df[col].max()

        min_IQR = Q1 - 1.5 * IQR
        max_IQR = Q3 + 1.5 * IQR

        print(f"\nAttribute - {col} : {df[col].dtypes}")
        print("Q1:", Q1, "| Q3:", Q3, "| IQR:", IQR)
        if col_min < min_IQR:
            print(f"  Low outlier threshold < {min_IQR}")
        if col_max > max_IQR:
            print(f"  High outlier threshold > {max_IQR}")
```

### 3.2 Opsi A – Drop baris outlier

```python
df_clean = df.copy()

for col in df.columns:
    if df[col].dtypes in [np.int64, np.float64] and col != 'symboling':
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        min_IQR = Q1 - 1.5 * IQR
        max_IQR = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[col] >= min_IQR) & (df_clean[col] <= max_IQR)]

print("Data awal:", df.shape)
print("Data setelah outlier dibuang:", df_clean.shape)
```

**Interpretasi:** Menghapus semua baris yang punya outlier di salah satu kolom numerik membuat data susut cukup banyak. Makin ketat threshold, makin banyak data yang dianggap outlier dan hilang.

### 3.3 Opsi B – Cek skewness, lalu tangani sesuai kondisi (median untuk skewed, winsorizing untuk mendekati normal)

```python
exclude_col = ['symboling', 'price']
for col in df.columns:
    if df[col].dtypes in [np.int64, np.float64] and col not in exclude_col:
        print(f"Attribute - {col} : skewness = {df[col].skew():.3f}")
```

```python
df_clean2 = df.copy()
exclude_col = ['symboling', 'price']

for col in df.columns:
    if df[col].dtypes in [np.int64, np.float64] and col not in exclude_col:
        skew_val = df[col].skew()

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        min_IQR = Q1 - 1.5 * IQR
        max_IQR = Q3 + 1.5 * IQR

        print(f"\nAttribute - {col}")
        print(f"Skewness: {skew_val:.3f}")

        if skew_val > 1 or skew_val < -1:
            median = df_clean2[col].median()
            df_clean2[col] = np.where((df_clean2[col] < min_IQR) | (df_clean2[col] > max_IQR),
                                       median, df_clean2[col])
            print("Outlier diganti dengan median.")
        else:
            df_clean2[col] = np.clip(df_clean2[col], min_IQR, max_IQR)
            print("Outlier ditangani dengan Winsorizing (clipping).")
```

**Interpretasi:** Pendekatan ini lebih halus dibanding drop baris: nilai outlier disesuaikan (di-clip ke batas IQR, atau diganti median untuk kolom yang sangat skewed), sehingga jumlah data tetap sama tapi pengaruh nilai ekstrem berkurang.

## 4. Data Formatting

### 4.1 Scaling data (Min-Max Scaler & Standard Scaler)

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

numeric_cols = ['horsepower', 'price', 'city-mpg']

scaler_minmax = MinMaxScaler()
scaler_std = StandardScaler()

df_minmax = pd.DataFrame(scaler_minmax.fit_transform(df_clean2[numeric_cols]), columns=numeric_cols)
df_std = pd.DataFrame(scaler_std.fit_transform(df_clean2[numeric_cols]), columns=numeric_cols)

print("Data asli:")
display(df_clean2[numeric_cols].head())
print("\nMin-Max Scaling (0-1):")
display(df_minmax.head())
print("\nStandardization (Z-score):")
display(df_std.head())
```

**Interpretasi:** Min-Max Scaling memampatkan semua nilai ke rentang 0–1, cocok untuk membandingkan variabel dengan skala jauh berbeda. Standardization (Z-score) mengubah data agar berpusat di 0 dengan simpangan baku 1.

### 4.2 Categorical Encoding

- **Label Encoding** untuk atribut biner/ordinal, contoh: `num-of-doors`
- **One-hot Encoding** untuk atribut multi-kategori

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_clean2['num-of-doors-encoded'] = le.fit_transform(df_clean2['num-of-doors'])
df_clean2[['num-of-doors', 'num-of-doors-encoded']].drop_duplicates()
```

```python
df_encoded = pd.get_dummies(df_clean2, columns=['body-style'], dtype=int)
[c for c in df_encoded.columns if 'body-style' in c]
```

**Interpretasi:** `num-of-doors` hanya punya 2 kategori (`two`, `four`) sehingga cukup dengan Label Encoding. `body-style` punya lebih dari 2 kategori sehingga One-hot Encoding lebih tepat, supaya model tidak salah menafsirkan kategori tersebut sebagai punya urutan/peringkat.

## 5. Kesimpulan
